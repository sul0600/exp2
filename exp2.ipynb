## 1. Import libraries

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

## 2. Load the dataset

```python
df = pd.read_csv("train_u6lujuX_CVtuZ9i.csv")
df.head()
```

## 3. Understand the dataset

```python
print("Shape:", df.shape)
df.info()
```

```python
df.describe()
```

```python
# Check missing values
df.isna().sum()
```

## 4. Remove unnecessary column

```python
# Loan_ID is only an ID, so remove it
df = df.drop("Loan_ID", axis=1)

df.head()
```

## 5. Separate numerical and categorical columns

Numerical columns contain numbers. Categorical columns contain groups or names.

```python
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(include="object").columns

print("Numerical columns:", list(num_cols))
print("Categorical columns:", list(cat_cols))
```

## 6. Handle missing values

### Example 1: Fill numerical missing values with mean

This follows the `SimpleImputer()` method used in your exp2 notebook.

```python
imp = SimpleImputer(strategy="mean")
df[num_cols] = imp.fit_transform(df[num_cols])

df.isna().sum()
```

### Example 2: Fill categorical missing values with mode

```python
imp2 = SimpleImputer(strategy="most_frequent")
df[cat_cols] = imp2.fit_transform(df[cat_cols])

df.isna().sum()
```

## 7. Encode categorical data

### Example 3: One-hot encoding using get_dummies()

`get_dummies()` converts categories into columns containing 0 and 1. This is the simple Pandas method also shown in your exp2 notebook.

```python
# Convert Loan_Status: Y = 1 and N = 0
df["Loan_Status"] = df["Loan_Status"].map({"Y": 1, "N": 0})

# Convert other categorical columns into 0 and 1 columns
df = pd.get_dummies(df, drop_first=True, dtype=int)

df.head()
```

## 8. Feature scaling

### Example 4: StandardScaler

StandardScaler changes numerical features to a similar scale. It uses mean and standard deviation.

```python
# Separate features and target
X = df.drop("Loan_Status", axis=1)
y = df["Loan_Status"]

sc = StandardScaler()
df_scaled = pd.DataFrame(sc.fit_transform(X), columns=X.columns)

# Add target column without scaling it
df_scaled["Loan_Status"] = y

df_scaled.head()
```

### Example 5: MinMaxScaler (another scaling method)

```python
df_minmax = pd.DataFrame(MinMaxScaler().fit_transform(X), columns=X.columns)
df_minmax["Loan_Status"] = y

df_minmax.head()
```

## 9. Check final preprocessed dataset

```python
print("Final shape:", df_scaled.shape)
print("Missing values:", df_scaled.isna().sum().sum())

df_scaled.head()
```

## 10. Conclusion

Thus, data preprocessing was performed on the Loan Approval dataset. Missing values were handled using SimpleImputer, categorical data was encoded using get_dummies(), and feature scaling was performed using StandardScaler and MinMaxScaler. The dataset is now in a numerical and scaled format.
